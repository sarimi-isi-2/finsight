```python
import pandas as pd
```

```python
# baca seluruh dataset
df1 = pd.read_csv("../data/raw/train_transaction_dataset.csv")
df2 = pd.read_csv("../data/raw/test_transaction_dataset.csv")
df3 = pd.read_csv("../data/raw/Data_Finance_6_Bulan.csv")
df4 = pd.read_csv("../data/raw/transactions_rows.csv")
```

```python
# melihat ukuran dataset
print(df1.shape)
print(df2.shape)
print(df3.shape)
print(df4.shape)
```

```python
# melihat nama kolom setiap dataset
print(df1.columns)
print(df2.columns)
print(df3.columns)
print(df4.columns)
```

```python
# menampilkan 5 data teratas dari setiap dataset 
display(df1.head())

display(df2.head())

display(df3.head())

display(df4.head())
```

```python
# mengetahui informasi dataset
df1.info()
print("\n")
df2.info()
print("\n")
df3.info()
print("\n")
df4.info()
```

```python
# memeriksa data yang kosong atau tidak memiliki nilai
print(df1.isnull().sum())
print("\n")
print(df2.isnull().sum())
print("\n")
print(df3.isnull().sum())
print("\n")
print(df4.isnull().sum())
```

```python
# memeriksa data yang muncul lebih dari satu kali
print(df1.duplicated().sum())
print("\n")
print(df2.duplicated().sum())
print("\n")
print(df3.duplicated().sum())
print("\n")
print(df4.duplicated().sum())
```

```python
print("Missing Value")
print(df4.isnull().sum())

print("\nDuplicate")
print(df4.duplicated().sum())
```

```python
df4 = df4.dropna(subset=["description", "category_id"])
```

```python
df4 = df4.drop_duplicates()
```

```python
df4 = df4.reset_index(drop=True)
```

```python
print("Missing Value")
print(df4.isnull().sum())

print("\nDuplicate")
print(df4.duplicated().sum())
```
