```python
import pandas as pd
import numpy as np
import re
```

```python
df1 = pd.read_csv("../data/raw/csv/train_transaction_dataset.csv")
df2 = pd.read_csv("../data/raw/csv/test_transaction_dataset.csv")
df3 = pd.read_csv("../data/experimental/personal_finance_dataset_8000_extended.csv")
df4 = pd.read_csv("../data/experimental/Data_Finance_6_Bulan.csv")
```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
print(df4.shape)
```

```python
df3 = df3.rename(columns={
    "Description": "text",
    "Category": "label"
})
```

```python
df3 = df3[["text", "label"]]
```

```python
df4 = df4.rename(columns={
    "Title": "text",
    "Category": "label"
})
```

```python
df4 = df4[["text", "label"]]
```

```python
print(df1.columns)
print(df2.columns)
print(df3.columns)
print(df4.columns)
```

```python
df1["label"] = df1["label"].str.lower()
df2["label"] = df2["label"].str.lower()
df3["label"] = df3["label"].str.lower()
df4["label"] = df4["label"].str.lower()
```

```python
label_mapping = {
    # Food
    "food": "food",
    "grocery": "food",
    "makan & minum": "food",

    # Shopping
    "shopping": "shopping",
    "online shopping": "shopping",
    "electronics": "shopping",
    "clothing": "shopping",
    "belanja": "shopping",

    # Travel & Transport
    "travel": "travel",
    "transport": "transport",

    # Bills
    "bills": "bills",
    "tagihan": "bills",

    # Entertainment
    "entertainment": "entertainment",
    "hiburan": "entertainment",

    # Healthcare
    "healthcare": "healthcare",

    # Income
    "gaji": "income",

    # Investment
    "investment": "investment",
    "goals": "investment",

    # EMI
    "emi": "EMI",

    # Uncategorized
    "uncategorized": "other"
}
```

```python
df1["label"] = df1["label"].replace(label_mapping)
df2["label"] = df2["label"].replace(label_mapping)
df3["label"] = df3["label"].replace(label_mapping)
df4["label"] = df4["label"].replace(label_mapping)
```

```python
print(df1["label"].unique())
print(df2["label"].unique())
print(df3["label"].unique())
print(df4["label"].unique())
```

```python
df = pd.concat(
    [df1, df2, df3, df4],
    ignore_index=True
)
```

```python
import re

def clean_text(text):
    text = str(text).lower()

    text = re.sub(r'http\S+', '', text)

    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
df["clean_text"] = df["text"].apply(clean_text)
```
